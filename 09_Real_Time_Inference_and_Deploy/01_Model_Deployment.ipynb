# Μάθημα 9: Real-Time Inference (Αποθήκευση και Χρήση Μοντέλου)

Εκπαιδεύσαμε ένα μοντέλο. Και τώρα τι; 
Στον πραγματικό κόσμο, δεν εκπαιδεύουμε το μοντέλο κάθε φορά που έχουμε νέα δεδομένα. Αντίθετα, **αποθηκεύουμε (σώζουμε)** το εκπαιδευμένο μοντέλο σε ένα αρχείο, και το 'φορτώνουμε' σε μια web εφαρμογή (π.χ. Django/Flask/Streamlit) για να κάνει προβλέψεις (Inference) σε πραγματικό χρόνο.

```python
import joblib
from sklearn.linear_model import LogisticRegression
import numpy as np

# 1. Έστω ότι έχουμε ένα απλό εκπαιδευμένο μοντέλο (το εκπαιδεύουμε γρήγορα εδώ)
X_train = np.array([[1, 2], [2, 3], [5, 5], [6, 7]])
y_train = np.array([0, 0, 1, 1])
model = LogisticRegression()
model.fit(X_train, y_train)

print("Το μοντέλο εκπαιδεύτηκε!")
```

## Αποθήκευση του Μοντέλου (Serialization)
Χρησιμοποιούμε τη βιβλιοθήκη `joblib` (ή `pickle`) για να αποθηκεύσουμε το μοντέλο στο δίσκο μας ως αρχείο `.pkl`.

```python
# Αποθήκευση...
joblib.dump(model, 'my_trained_model.pkl')
print("Το μοντέλο αποθηκεύτηκε στο αρχείο 'my_trained_model.pkl'!")
```

## Φόρτωση και Πρόβλεψη σε Πραγματικό Χρόνο
Τώρα, ας πούμε ότι βρισκόμαστε σε ένα άλλο script (σε έναν web server). Ο χρήστης δίνει νέα δεδομένα.

```python
# Φόρτωση του μοντέλου από το αρχείο
loaded_model = joblib.load('my_trained_model.pkl')

# Νέα δεδομένα από τον χρήστη (π.χ. από μια φόρμα στο ίντερνετ)
new_user_data = np.array([[3, 4]])

# Πρόβλεψη (Inference)
prediction = loaded_model.predict(new_user_data)
print("Το μοντέλο προέβλεψε την κατηγορία:", prediction[0])
```
