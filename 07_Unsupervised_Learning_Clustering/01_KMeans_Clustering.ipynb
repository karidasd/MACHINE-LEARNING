# Μάθημα 7: Unsupervised Learning - Clustering (Ομαδοποίηση)

Στη Μηχανική Μάθηση **Χωρίς Επίβλεψη (Unsupervised Learning)**, δίνουμε στο μοντέλο μόνο τα δεδομένα ($X$) ΧΩΡΙΣ ετικέτες/αποτελέσματα ($y$). Το αφήνουμε να βρει μόνο του μοτίβα. 

Το πιο κοινό πρόβλημα είναι το **Clustering (Ομαδοποίηση)**: Προσπαθούμε να χωρίσουμε τα δεδομένα σε ομάδες που μοιάζουν μεταξύ τους (π.χ. Customer Segmentation για marketing).

## Ο αλγόριθμος K-Means
Ο K-Means προσπαθεί να βρει 'K' κέντρα βάρους (centroids) στα δεδομένα και να αναθέσει κάθε σημείο στο πιο κοντινό του κέντρο.

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

# 1. Δημιουργία Συνθετικών Δεδομένων (4 ξεκάθαρες ομάδες)
X, y_true = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)

plt.scatter(X[:, 0], X[:, 1], s=50)
plt.title("Ακατέργαστα Δεδομένα (Χωρίς ετικέτες)")
plt.show()
```

## Εκπαίδευση του K-Means
Θα ζητήσουμε από τον αλγόριθμο να βρει 4 ομάδες (clusters).

```python
# Αρχικοποίηση και εκπαίδευση (ΠΡΟΣΟΧΗ: Δίνουμε ΜΟΝΟ το X)
kmeans = KMeans(n_clusters=4, random_state=42)
kmeans.fit(X)

# Ποιες είναι οι ετικέτες που 'εφηύρε' ο αλγόριθμος για κάθε σημείο;
predicted_clusters = kmeans.labels_

# Πού βρίσκονται τα κέντρα των ομάδων;
centers = kmeans.cluster_centers_

# Οπτικοποίηση Αποτελεσμάτων
plt.scatter(X[:, 0], X[:, 1], c=predicted_clusters, s=50, cmap='viridis')
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.7, marker='X') # Ζωγραφίζουμε τα κέντρα με κόκκινο Χ
plt.title("Ομαδοποίηση (Clustering) με K-Means")
plt.show()
```
